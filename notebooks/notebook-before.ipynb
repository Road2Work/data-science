### Cleaning Data

#### Fixing Duplicate Data, Missing Value, Format Standarization, Outlier problem

```python
print("=== Start Data Cleaning ===")
df_clean = df.copy()

# 1. Drop duplikat
df_clean = df_clean.drop_duplicates()
print(f"Drop duplikat -> sisa: {df_clean.shape[0]} baris")

# 2. Drop kolom zero-variance (domain)
df_clean = df_clean.drop(columns=['domain'], errors='ignore')
print("Drop kolom 'domain' selesai")

# 3. Drop NaN di kolom skor kritikal
kolom_kritikal = [
    'answer', 'star_structure', 'role_relevance', 
    'communication_clarity', 'technical_accuracy', 'evidence_specificity', 'self_awareness'
]
df_clean = df_clean.dropna(subset=kolom_kritikal)
print(f"Drop NaN kritikal -> sisa: {df_clean.shape[0]} baris")

# 4. Imputasi weakness_tags
df_clean['weakness_tags'] = df_clean['weakness_tags'].fillna('no_weakness')
print("Imputasi 'weakness_tags' -> 'no_weakness'")

# 5. Rule-based imputation untuk quality_label
def isi_quality_label(row):
    if pd.isna(row['quality_label']):
        if row['final_score_0_1'] >= 0.75: return 'Strong'
        elif row['final_score_0_1'] >= 0.50: return 'Average'
        else: return 'Weak'
    return row['quality_label']

df_clean['quality_label'] = df_clean.apply(isi_quality_label, axis=1)

# Standardisasi format (Title Case)
df_clean['quality_label'] = df_clean['quality_label'].astype(str).str.strip().str.title()
print("Imputasi & standardisasi 'quality_label' selesai")

# 6. Final drop jika masih ada label aneh
valid_labels = ['Strong', 'Average', 'Weak']
df_clean = df_clean[df_clean['quality_label'].isin(valid_labels)]

# Validasi Akhir
print("\n[+] Cek missing values akhir:")
print(df_clean.isna().sum()[df_clean.isna().sum() > 0]) # Cuma nge-print yang masih ada NaN

print(f"\n[+] Cleaning beres! Dimensi final: {df_clean.shape}")
```

```python
df_clean.head()
```

**Insight:**
- Proses drop duplicates dan pembersihan missing values pada fitur krusial menyusutkan dataset dari 5.339 menjadi 5.168 baris. Penyusutan sekitar ~3% ini sangat aman dan menjaga integritas data secara keseluruhan.

- Kolom domain berhasil di-drop karena terbukti tidak memiliki variansi (zero-variance).

- Strategi rule-based imputation sukses menyelamatkan baris yang kehilangan label kualitas tanpa harus merusak struktur distribusi nilai awalnya.

- Standardization dengan str.title() berhasil meratakan anomali kapitalisasi ('STRONG' atau 'average'), sehingga kategori pada quality_label kini kembali valid menjadi tiga kelas bersih: Weak, Average, dan Strong.

## Feature Engineering

Pada tahap ini, kita akan mengekstrak metrik baru dari teks jawaban (`answer`) kandidat. Fitur-fitur ini digunakan untuk membedah anatomi jawaban ideal, mengukur panjang teks, serta mendeteksi komponen struktural STAR (Situation, Task, Action, Result) dan bukti numerik guna menjawab pertanyaan bisnis nomor 3 dan 4.

```python
import re

print("=== Start Feature Engineering ===")
df_features = df_clean.copy()

# 1. Hitung jumlah kata (Verbosity)
df_features['answer_length_words'] = df_features['answer'].apply(lambda x: len(str(x).split()))
print("[+] Ekstraksi 'answer_length_words' selesai")

# 2. Ekstraksi metrik kelengkapan umum STAR (1 = aman, 0 = ada missing)
df_features['has_context'] = df_features['weakness_tags'].apply(lambda x: 0 if 'context_missing' in str(x) else 1)
df_features['has_contribution'] = df_features['weakness_tags'].apply(lambda x: 0 if 'contribution_unclear' in str(x) else 1)
df_features['has_tools'] = df_features['weakness_tags'].apply(lambda x: 0 if 'tools_missing' in str(x) else 1)
df_features['has_impact'] = df_features['weakness_tags'].apply(lambda x: 0 if 'impact_missing' in str(x) else 1)
df_features['has_metric'] = df_features['weakness_tags'].apply(lambda x: 0 if 'measurable_result_missing' in str(x) else 1)
df_features['has_technical_detail'] = df_features['weakness_tags'].apply(lambda x: 0 if 'technical_detail_missing' in str(x) else 1)
df_features['has_self_awareness'] = df_features['weakness_tags'].apply(lambda x: 0 if 'self_awareness_missing' in str(x) else 1)
print("[+] Ekstraksi komponen biner STAR selesai")

# 3. Hitung Filler Words (Tanda keraguan komunikasi)
def count_fillers(text):
    fillers = r'\b(ehh|emm|hmm|kayaknya|anu|ya|gitu|kayak)\b'
    return len(re.findall(fillers, str(text).lower()))

df_features['filler_word_count'] = df_features['answer'].apply(count_fillers)
print("[+] Ekstraksi 'filler_word_count' selesai")

# 4. Ekstraksi metrik spesifik berdasarkan target role (Role-Aware Logic)
def check_testing_coverage(row):
    if row['target_role'] in ['DevOps Engineer', 'Manual QA Tester', 'QA Automation Engineer', 'Software Tester']:
        return 0 if 'testing_coverage_missing' in str(row['weakness_tags']) else 1
    return None

def check_deployment_detail(row):
    if row['target_role'] in ['Cloud Engineer', 'DevOps Engineer', 'QA Automation Engineer']:
        return 0 if 'deployment_missing' in str(row['weakness_tags']) else 1
    return None

def check_mobile_detail(row):
    if row['target_role'] in ['Mobile Developer', 'Android Developer', 'iOS Developer']:
        return 0 if 'mobile_platform_detail_missing' in str(row['weakness_tags']) else 1
    return None

def check_security_awareness(row):
    if row['target_role'] in ['Cloud Engineer', 'DevOps Engineer', 'Cybersecurity Analyst']:
        return 0 if 'security_awareness_missing' in str(row['weakness_tags']) else 1
    return None

def check_backend_architecture(row):
    if row['target_role'] in ['Backend Developer', 'Software Engineer']:
        return 0 if 'backend_architecture_unclear' in str(row['weakness_tags']) else 1
    return None

def check_model_evaluation(row):
    if row['target_role'] in ['ML Engineer', 'Data Scientist', 'AI Engineer']:
        return 0 if 'model_evaluation_missing' in str(row['weakness_tags']) else 1
    return None

def check_data_processing(row):
    if row['target_role'] in ['Data Scientist', 'Data Analyst', 'Data Engineer', 'ML Engineer']:
        return 0 if 'data_processing_unclear' in str(row['weakness_tags']) else 1
    return None

df_features['has_testing_coverage'] = df_features.apply(check_testing_coverage, axis=1)
df_features['has_deployment'] = df_features.apply(check_deployment_detail, axis=1)
df_features['has_mobile_platform_detail'] = df_features.apply(check_mobile_detail, axis=1)
df_features['has_security_awareness'] = df_features.apply(check_security_awareness, axis=1)
df_features['has_backend_architecture'] = df_features.apply(check_backend_architecture, axis=1)
df_features['has_model_evaluation'] = df_features.apply(check_model_evaluation, axis=1)
df_features['has_data_processing'] = df_features.apply(check_data_processing, axis=1)
print("[+] Ekstraksi metrik Role-Aware selesai\n")

# Cek hasil akhir
kolom_tampil = ['target_role', 'answer_length_words', 'filler_word_count', 'has_context', 'has_metric', 'quality_label', 'has_model_evaluation']
print("=== Preview Hasil Feature Engineering (Sample) ===")
display(df_features[kolom_tampil].head())
```

Insight:

- Ekstraksi answer_length_words dan filler_word_count berhasil dilakukan. Dua fitur numerik ini akan menjadi kunci utama dalam EDA untuk mendeteksi verbosity bias (apakah jawaban panjang pasti nilainya bagus?) dan memvalidasi penalti terhadap kandidat yang terlalu sering menggunakan kata pengisi lisan.

- Kolom-kolom weakness_tags yang sebelumnya bertumpuk (berupa teks panjang) berhasil dipecah (One-Hot Encoding biner) menjadi metrik struktural seperti has_context, has_metric, dll. Nilai 1 berarti kandidat sukses melengkapi komponen tersebut, sedangkan 0 berarti gagal/absen.

- Strategi Role-Aware Logic berjalan sesuai ekspektasi. Perhatikan kolom seperti has_model_evaluation; fitur ini menghasilkan nilai logis (1 atau 0) hanya untuk rumpun AI/ML, namun otomatis menghasilkan nilai NaN (kosong) untuk peran di luar rumpun tersebut (seperti QA Tester) karena kompetensi tersebut tidak relevan untuk diuji.

**Mengapa Pertanyaan Ini Penting?**
Analisis ini digunakan untuk mendeteksi rumpun peran (*Role Family*) mana yang paling kritis atau paling banyak diisi oleh kandidat "omong doang" (kandidat yang hanya berada di *Evidence Level* 1 atau 2). Hasil ini penting agar platform Road2Work bisa menargetkan distribusi modul remedial pembuatan portofolio secara tepat sasaran di pengembangan berikutnya.

Informasi tiga kelemahan teratas ini akan menjadi fondasi utama dalam menyusun algoritma *Clarifying Question Engine* agar bisa menembak kelemahan user secara otomatis.

### Pertanyaan 7: Bagaimana korelasi antara tingkat relevansi jawaban kandidat terhadap peran yang dituju dengan pencapaian tingkat kedalaman bukti konkret yang mereka berikan?

**Mengapa Pertanyaan Ini Penting?**
Analisis korelasi ini bertujuan menguji kebenaran hipotesis rekrutmen: *"Apakah kandidat yang bicaranya sangat nyambung dengan posisi lamaran (role_relevance tinggi) otomatis mampu memberikan bukti kerja yang mendalam (evidence_level tinggi)?"*. Jika korelasinya rendah, berarti banyak pelamar yang sekadar pintar mencocokkan kata kunci (*keyword matching*) tanpa benar-benar punya pengalaman nyata.

```python
# Menghitung korelasi Spearman antara skor relevansi peran dengan tingkatan level bukti
# Pakai Spearman karena evidence_level sifatnya data ordinal/tingkatan (1 sampai 5)
korelasi_spearman = df_eda['role_relevance'].corr(df_eda['evidence_level'], method='spearman')

print(f"Koefisien Korelasi Spearman antara Relevansi Peran dan Level Bukti: {korelasi_spearman:.4f}")

# Tambahan analisis: tampilkan rata-rata nilai relevansi untuk setiap tingkatan evidence_level
print("\nRata-rata Skor Relevansi Peran pada Tiap Tingkat Kedalaman Bukti (Evidence Level 1-5):")
print(df_eda.groupby('evidence_level')['role_relevance'].mean().reset_index().to_string(index=False))
```

### Pertanyaan 8: Apakah tingginya frekuensi penggunaan kata pengisi lisan (*filler words*) berkorelasi dengan rendahnya kualitas jawaban secara keseluruhan?

**Mengapa Pertanyaan Ini Penting?**
Di dunia nyata, HRD sangat memperhatikan kelancaran komunikasi. Terlalu banyak *filler words* (seperti 'ehh', 'emm', 'kayaknya') sering kali menjadi sinyal keraguan, kurangnya rasa percaya diri, atau minimnya penguasaan materi teknis. Analisis ini bertujuan untuk membuktikan apakah sistem berhasil menangkap fenomena psikologis ini dan menghukum kandidat (masuk ke label *Weak*) yang terlalu banyak terbata-bata.

```python
# Menghitung rata-rata, median, dan jumlah maksimal filler words per kategori kualitas
pilar3_q8 = df_eda.groupby('quality_label')['filler_word_count'].agg(['mean', 'median', 'max']).reset_index()

# Urutkan dari Strong ke Weak agar lebih intuitif saat dibaca
kategori_urut = {'Strong': 0, 'Average': 1, 'Weak': 2}
pilar3_q8['sort_order'] = pilar3_q8['quality_label'].map(kategori_urut)
pilar3_q8 = pilar3_q8.sort_values(by='sort_order').drop(columns=['sort_order'])

# Ubah nama kolom agar enak dibaca saat di-print
pilar3_q8.columns = ['Kategori Kualitas', 'Rata-rata Filler', 'Median Filler', 'Filler Terbanyak (Max)']

print("=== Statistik Penggunaan Filler Words Lintas Kategori ===")
print(pilar3_q8.to_string(index=False))

# Tambahan: Cek berapa persen kandidat Strong yang benar-benar bersih (0 filler words)
bersih_strong = len(df_eda[(df_eda['quality_label'] == 'Strong') & (df_eda['filler_word_count'] == 0)])
total_strong = len(df_eda[df_eda['quality_label'] == 'Strong'])
persentase_bersih = (bersih_strong / total_strong) * 100 if total_strong > 0 else 0

print(f"\nInsight Tambahan: {persentase_bersih:.1f}% kandidat di kategori 'Strong' berbicara sangat lancar tanpa filler words sama sekali.")
```

### A/B Testing: Signifikansi Penggunaan Metrik Angka Terhadap Skor Akhir

```python
print("="*60)
print("=== EKSPERIMEN A/B TESTING & UJI HIPOTESIS STATISTIK ===")
print("="*60)

# -------------------------------------------------------------------
# EKSPERIMEN 1: PENGARUH METRIK ANGKA (has_metric)
# -------------------------------------------------------------------
print("\n[1] HIPOTESIS: Penggunaan metrik angka (has_metric) memengaruhi skor akhir secara signifikan.")
g1_metric = df_eda[df_eda['has_metric'] == 1]['final_score_0_100'].dropna()
g1_nometric = df_eda[df_eda['has_metric'] == 0]['final_score_0_100'].dropna()

t_stat1, p_val1 = stats.ttest_ind(g1_metric, g1_nometric, equal_var=False)

print(f"    Rata-rata Skor (Ada Metrik)   : {g1_metric.mean():.2f}")
print(f"    Rata-rata Skor (Tanpa Metrik) : {g1_nometric.mean():.2f}")
print(f"    P-Value                       : {p_val1:.5e}")
if p_val1 < 0.05:
    print("    => KESIMPULAN: SIGNIFIKAN (Hipotesis Diterima).")
else:
    print("    => KESIMPULAN: TIDAK SIGNIFIKAN.")


# -------------------------------------------------------------------
# EKSPERIMEN 2: PENGARUH DAMPAK BISNIS (has_impact)
# -------------------------------------------------------------------
print("\n[2] HIPOTESIS: Menyebutkan dampak bisnis (has_impact) memengaruhi skor akhir secara signifikan.")
g2_impact = df_eda[df_eda['has_impact'] == 1]['final_score_0_100'].dropna()
g2_noimpact = df_eda[df_eda['has_impact'] == 0]['final_score_0_100'].dropna()

t_stat2, p_val2 = stats.ttest_ind(g2_impact, g2_noimpact, equal_var=False)

print(f"    Rata-rata Skor (Ada Impact)   : {g2_impact.mean():.2f}")
print(f"    Rata-rata Skor (Tanpa Impact) : {g2_noimpact.mean():.2f}")
print(f"    P-Value                       : {p_val2:.5e}")
if p_val2 < 0.05:
    print("    => KESIMPULAN: SIGNIFIKAN (Hipotesis Diterima).")
else:
    print("    => KESIMPULAN: TIDAK SIGNIFIKAN.")


# -------------------------------------------------------------------
# EKSPERIMEN 3: PENGARUH FILLER WORDS TERHADAP SKOR KOMUNIKASI
# -------------------------------------------------------------------
print("\n[3] HIPOTESIS: Kandidat dengan Filler Words tinggi memiliki skor Komunikasi (communication_clarity) yang lebih rendah.")
# Kita bagi 2 kelompok: High Filler (> median) vs Low Filler (<= median)
median_filler = df_eda['filler_word_count'].median()
g3_high_filler = df_eda[df_eda['filler_word_count'] > median_filler]['communication_clarity'].dropna()
g3_low_filler = df_eda[df_eda['filler_word_count'] <= median_filler]['communication_clarity'].dropna()

t_stat3, p_val3 = stats.ttest_ind(g3_low_filler, g3_high_filler, equal_var=False)

print(f"    Batas Median Filler Words           : {median_filler} kata")
print(f"    Skor Komunikasi (Low Filler <= {median_filler})  : {g3_low_filler.mean():.2f}")
print(f"    Skor Komunikasi (High Filler > {median_filler})  : {g3_high_filler.mean():.2f}")
print(f"    P-Value                             : {p_val3:.5e}")
if p_val3 < 0.05:
    print("    => KESIMPULAN: SIGNIFIKAN (Hipotesis Diterima).")
else:
    print("    => KESIMPULAN: TIDAK SIGNIFIKAN.")

# -------------------------------------------------------------------
# EKSPERIMEN 4: VERBOSITY BIAS (PENGARUH PANJANG JAWABAN) - Terkait Q4
# -------------------------------------------------------------------
print("\n[4] HIPOTESIS: Jawaban yang sangat panjang (Verbose > 250 kata) mendapatkan skor teknis yang secara signifikan berbeda dari jawaban padat (Concise <= 250 kata).")
# Membagi grup berdasarkan titik jenuh dari Pertanyaan 4
g4_verbose = df_eda[df_eda['answer_length_words'] > 250]['technical_accuracy'].dropna()
g4_concise = df_eda[df_eda['answer_length_words'] <= 250]['technical_accuracy'].dropna()

t_stat4, p_val4 = stats.ttest_ind(g4_verbose, g4_concise, equal_var=False)

print(f"    Rata-rata Skor Teknis (Verbose > 250) : {g4_verbose.mean():.2f}")
print(f"    Rata-rata Skor Teknis (Concise <=250) : {g4_concise.mean():.2f}")
print(f"    P-Value                               : {p_val4:.5e}")
if p_val4 < 0.05:
    print("    => KESIMPULAN: SIGNIFIKAN. Panjang kata memang memengaruhi penilaian secara statistik (Waspada Verbosity Bias!).")
else:
    print("    => KESIMPULAN: TIDAK SIGNIFIKAN. Kualitas substansi lebih menentukan daripada sekadar jumlah kata.")


# -------------------------------------------------------------------
# EKSPERIMEN 5: PENALTI KLARIFIKASI (need_clarification) - Terkait Q6
# -------------------------------------------------------------------
print("\n[5] HIPOTESIS: Kandidat yang membutuhkan pertanyaan klarifikasi lanjutan (need_clarification = True) memiliki skor akhir yang secara signifikan lebih rendah.")
g5_need_clarify = df_eda[df_eda['need_clarification'] == True]['final_score_0_100'].dropna()
g5_no_clarify = df_eda[df_eda['need_clarification'] == False]['final_score_0_100'].dropna()

t_stat5, p_val5 = stats.ttest_ind(g5_need_clarify, g5_no_clarify, equal_var=False)

print(f"    Rata-rata Skor (Butuh Klarifikasi)  : {g5_need_clarify.mean():.2f}")
print(f"    Rata-rata Skor (Lancar/Tanpa Tanya) : {g5_no_clarify.mean():.2f}")
print(f"    P-Value                             : {p_val5:.5e}")
if p_val5 < 0.05:
    print("    => KESIMPULAN: SIGNIFIKAN (Hipotesis Diterima). Membutuhkan klarifikasi sangat merugikan nilai akhir kandidat.")
else:
    print("    => KESIMPULAN: TIDAK SIGNIFIKAN.")


# -------------------------------------------------------------------
# EKSPERIMEN 6: PENGARUH PEMBUKAAN KONTEKS (has_context) - Terkait Q3 & Q7
# -------------------------------------------------------------------
print("\n[6] HIPOTESIS: Memberikan latar belakang/konteks (has_context = 1) di awal jawaban akan secara signifikan meningkatkan skor Relevansi Peran (role_relevance).")
g6_with_context = df_eda[df_eda['has_context'] == 1]['role_relevance'].dropna()
g6_no_context = df_eda[df_eda['has_context'] == 0]['role_relevance'].dropna()

t_stat6, p_val6 = stats.ttest_ind(g6_with_context, g6_no_context, equal_var=False)

print(f"    Rata-rata Skor Relevansi (Ada Konteks)   : {g6_with_context.mean():.2f}")
print(f"    Rata-rata Skor Relevansi (Tanpa Konteks) : {g6_no_context.mean():.2f}")
print(f"    P-Value                                  : {p_val6:.5e}")
if p_val6 < 0.05:
    print("    => KESIMPULAN: SIGNIFIKAN (Hipotesis Diterima). Latar belakang sangat penting agar jawaban terasa nyambung dengan posisi.")
else:
    print("    => KESIMPULAN: TIDAK SIGNIFIKAN.")
```

## Data Splitting (Train, Validation, Test Split)
Pada tahap ini, dataset yang telah bersih dan kaya fitur akan dibagi menjadi tiga bagian terpisah: Train Set (70%) untuk melatih model, Validation Set (15%) untuk tuning dan validasi internal, serta Test Set (15%) sebagai ujian akhir untuk mengevaluasi performa model Machine Learning secara objektif.

```python
# 1. Tentukan fitur (X) dan target (y) 
df_eda['stratify_col'] = df_eda['role_family'].astype(str) + "_" + df_eda['quality_label'].astype(str)

# Split pertama: Pisahkan Train (70%) dan Sisa/Temp (30%)
train_df, temp_df = train_test_split(
    df_eda, 
    test_size=0.30, 
    random_state=42, 
    stratify=df_eda['stratify_col']
)

# Split kedua: Bagi Sisa/Temp (30%) tadi menjadi Validation (15%) dan Test (15%) seimbang rata
val_df, test_df = train_test_split(
    temp_df, 
    test_size=0.50, 
    random_state=42, 
    stratify=temp_df['stratify_col']
)

# Drop kembali kolom pembantu stratify agar data tetap steril
for data_split in [train_df, val_df, test_df]:
    data_split.drop(columns=['stratify_col'], inplace=True, errors='ignore')

# 2. Definisikan path folder tujuan di kasta tertinggi (03_processed)
processed_dir = "../data/03_processed/"
os.makedirs(processed_dir, exist_ok=True)

# Ekspor ketiga pecahan data menjadi file CSV terpisah
train_df.to_csv(os.path.join(processed_dir, "train_df.csv"), index=False)
val_df.to_csv(os.path.join(processed_dir, "val_df.csv"), index=False)
test_df.to_csv(os.path.join(processed_dir, "test_df.csv"), index=False)

# 3. Print laporan hasil pemotongan data untuk validasi akhir
print("=== LAPORAN DISTRIBUSI DATA SPLITTING ===")
print(f"Total Data Awal : {len(df_eda)} baris")
print(f"Train Set (80%) : {len(train_df)} baris ➔ Disimpan di 03_processed/train_df.csv")
print(f"Val Set (10%)   : {len(val_df)} baris ➔ Disimpan di 03_processed/val_df.csv")
print(f"Test Set (10%)  : {len(test_df)} baris ➔ Disimpan di 03_processed/test_df.csv")
```
